```python
import numpy as np
import trimesh

def load_pointcloud_txt(path):
    """
    Load a point cloud from a TXT file with the format:
    x y z nx ny nz r g b
    """
    # Load all floats from file
    data = np.loadtxt(path)

    # Split into components
    points = data[:, 0:3]
    normals = data[:, 3:6]

    # Colors: convert from 0–255 to 0–1
    colors = data[:, 6:9] / 255.0

    # Create trimesh point cloud
    cloud = trimesh.points.PointCloud(
        vertices=points,
        colors=colors,
        normals=normals
    )

    return cloud

cloud = load_pointcloud_txt(pcdPath)
cloud._data.__dict__
```

```python
import open3d as o3d
import numpy as np

def load_pointcloud_txt_o3d(path):
    data = np.loadtxt(path)

    points = data[:, 0:3]
    normals = data[:, 3:6]
    colors = data[:, 6:9] / 255.0

    pcd = o3d.geometry.PointCloud()
    pcd.points = o3d.utility.Vector3dVector(points)
    pcd.normals = o3d.utility.Vector3dVector(normals)
    pcd.colors = o3d.utility.Vector3dVector(colors)

    return pcd

pcd = load_pointcloud_txt_o3d(pcdPath)

# Display inline
o3d.visualization.draw_geometries([pcd])
```

### Voxels

```python
import json
def load_voxel_json(path):
    """
    Parses a voxel JSON file with fields:
    - voxelSize
    - origin {x,y,z}
    - rotation {x,y,z,w}
    - voxels[] with:
        - gridIndex {x,y,z}
        - color {r,g,b,a}
        - distance
        - worldPos {x,y,z}
    """

    with open(path, "r") as f:
        data = json.load(f)

    voxel_size = data["voxelSize"]
    origin = np.array([
        data["origin"]["x"],
        data["origin"]["y"],
        data["origin"]["z"]
    ], dtype=float)

    rotation = np.array([
        data["rotation"]["x"],
        data["rotation"]["y"],
        data["rotation"]["z"],
        data["rotation"]["w"]
    ], dtype=float)

    voxels = data["voxels"]
    occludedVoxels = data["occludedVoxels"]

    # Voxels
    # Extract into arrays
    world_positions = []
    colors = []
    grid_indices = []
    distances = []

    for v in voxels:
        wp = v["worldPos"]
        col = v["color"]
        gi = v["gridIndex"]

        world_positions.append([wp["x"], wp["y"], wp["z"]])
        colors.append([col["r"], col["g"], col["b"], col["a"]])
        grid_indices.append([gi["x"], gi["y"], gi["z"]])
        distances.append(v["distance"])

    world_positions = np.array(world_positions, dtype=float)
    colors = np.array(colors, dtype=float)
    grid_indices = np.array(grid_indices, dtype=int)
    distances = np.array(distances, dtype=float)

    # OccludedVoxels
    # Extract into arrays
    occl_world_positions = []
    occl_colors = []
    occl_grid_indices = []
    occl_distances = []

    for v in occludedVoxels:
        wp = v["worldPos"]
        col = v["color"]
        gi = v["gridIndex"]

        occl_world_positions.append([wp["x"], wp["y"], wp["z"]])
        occl_colors.append([col["r"], col["g"], col["b"], col["a"]])
        occl_grid_indices.append([gi["x"], gi["y"], gi["z"]])
        occl_distances.append(v["distance"])

    occl_world_positions = np.array(occl_world_positions, dtype=float)
    occl_colors = np.array(occl_colors, dtype=float)
    occl_grid_indices = np.array(occl_grid_indices, dtype=int)
    occl_distances = np.array(occl_distances, dtype=float)

    return {
        "voxel_size": voxel_size,
        "origin": origin,
        "rotation": rotation,
        "world_positions": world_positions,
        "colors": colors,
        "grid_indices": grid_indices,
        "distances": distances,
        "occl_world_positions": occl_world_positions,
        "occl_colors": occl_colors,
        "occl_grid_indices": occl_grid_indices,
        "occl_distances": occl_distances,

    }
```

```python
voxels = load_voxel_json(voxelPath)
```

## Create a occupancy SDF

```python
import numpy as np

def build_voxel_grid(occupied_voxels, grid_size):
    """
    occupied_voxels: list of (x, y, z) integer indices
    grid_size: (nx, ny, nz)
    """
    grid = np.zeros(grid_size, dtype=np.uint8)
    for x, y, z in occupied_voxels:
        grid[x, y, z] = 1
    return grid

from skimage import measure

def marching_cubes_from_voxels(voxel_grid, voxel_size, origin):
    """
    voxel_grid: 3D numpy array with 1 = occupied, 0 = empty
    voxel_size: float (size of a voxel in world units)
    origin: world coordinate of voxel (0,0,0)
    """

    # marching cubes expects "values", so use the grid directly:
    verts, faces, normals, values = measure.marching_cubes(
        volume=voxel_grid,
        level=0.5,           # 0.5 = boundary of occupied voxels
        spacing=(voxel_size, voxel_size, voxel_size)
    )

    # shift vertices into world space
    verts += np.array(origin)

    return verts, faces, normals

def flip_normals(mesh):
    mesh.faces = mesh.faces[:, ::-1]
    return mesh
def clear_voxel_grid_edges(grid):
    """
    Sets all voxels on the outer faces of the grid to 0.
    Prevents marching cubes surfaces from touching the boundary,
    making the final mesh watertight.
    """
    nx, ny, nz = grid.shape

    # X faces
    grid[0, :, :] = 0
    grid[nx - 1, :, :] = 0

    # Y faces
    grid[:, 0, :] = 0
    grid[:, ny - 1, :] = 0

    # Z faces
    grid[:, :, 0] = 0
    grid[:, :, nz - 1] = 0

    return grid

```

```python
grid_size = (64,64,64)
occupied_voxels = voxels["occl_grid_indices"]

voxel_grid = build_voxel_grid(occupied_voxels, grid_size)
cleared_voxel_grid = clear_voxel_grid_edges(voxel_grid)
voxel_size = 0.32
origin = (-0.0268, 9.9774, -0.3420)

verts, faces, normals = marching_cubes_from_voxels(voxel_grid, voxel_size, origin)

mesh = trimesh.Trimesh(vertices=verts, faces=faces, vertex_normals=normals)
finalmesh = flip_normals(mesh)
```
